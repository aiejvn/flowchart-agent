```python
from utils import split_docs
import os

import json


data_dir = './constructive_dismissal/test'
out_file = 'cd.jsonl'
```

```python



data = []
for dir in os.listdir(data_dir):
    temp = split_docs(os.path.join(data_dir, dir))
    
    data.append({'query': 'Is there a case for constructive dismissal in this client\'s information?',
                 'supporting_docs': temp,
                 'gold_label': ''
                 })
```

```python
print(data[0]['supporting_docs'][0]['content'])
```

```python

with open(out_file, 'w', encoding='utf-8') as f:
    for d in data:
        f.write(json.dumps(d) + '\n')
```
