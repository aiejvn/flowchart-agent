```python
from utils import split_docs, process_pdf
import os

import json


data_dir = './usecases/constructive_dismissal/test'
caselaw_dir =  './usecases/constructive_dismissal/caselaw'
out_file = 'cd_synthetic.jsonl'
```

```python


fns = list(os.listdir(caselaw_dir))
fns_short = [x.split()[0].lower() for x in fns]



data = []
for dir in os.listdir(data_dir):
    print(data_dir, dir)
    gold_ans, supporting = split_docs(os.path.join(data_dir, dir))

    temp = dir.split()[0].lower()
    if temp in fns_short:
        case = process_pdf(caselaw_dir, fns[fns_short.index(temp)])
    
    data.append({'query': 'Is there a case for constructive dismissal in this client\'s information?',
                 'supporting_docs': supporting,
                 'original_case': '\n'.join([x for x in case if len(x) > 0]),
                 'gold_label': gold_ans
                 })
```
